```python
from openai import OpenAI
import base64
import os
from PIL import Image

openai_apikey = os.getenv('OPENAI_APIKEY')
openai_organization = os.getenv('OPENAI_ORGANIZATION')

NumIterations = 30

client = OpenAI(
    organization=openai_organization,
    api_key=openai_apikey,  
)
```

```python
def image_from_image(image_path):
    with open(image_path, "rb") as image_file:
        image_response = client.images.edit(
            image=image_file,
            model = "gpt-image-2",
            prompt="Please, please, please; provide me the same image as I'm uploading. Please keep track of all details on the image, the colors, the sizing, objects and more. "+ \
                "Check, double-check and triple-check if the provided image is indeed the same as the uploaded image.", # Please the AI model to make sure the Terminator doesn't come after me.
            size="1024x1024"
        )
    return image_response.data[0].b64_json
```

```python
for i in range(0, NumIterations):
    print("Start image ",i)
    gen_picture = image_from_image(f"./PictureLoop/image_{i}.png")
    with open(f"./PictureLoop/image_{i+1}.png", "wb") as f:
        f.write(base64.b64decode(gen_picture))
```

```python
output_gif = 'ImageEvolution.gif'

image_files = [f'image_{i}.png' for i in range(0,NumIterations)]
images = [Image.open(os.path.join('PictureLoop', img)) for img in image_files]

# Save as GIF
images[0].save(
    output_gif,
    save_all=True,
    append_images=images[1:],
    duration=500,   # duration per frame in milliseconds
    loop=0          # loop forever
)

print(f"GIF saved as {output_gif}")
```
